#Data Wrangling
**Combining Data using merge and join**

```python
import pandas as pd
students=pd.DataFrame({
    'Student_ID':[101,102,103,104],
    'Name':['Aman','Priya','Rahul','Shagun'],
    'Dept':["CSE","CSE","DS","ECE"]
})
marks=pd.DataFrame({
    'Student_ID':[101,102,104,105],
    'Marks':[87,72,91,86]
})
```

```python
print("Student table:")
print(students)
print("\nMarks table:")
print(marks)
```

```python
inner=pd.merge(students,marks,on='Student_ID',how='inner')
print("\nInner join:")
print(inner )
```

```python
left=pd.merge(students,marks,on='Student_ID',how='left')
print("\nLeft join:")
print(left)
```

```python
right=pd.merge(students,marks,on='Student_ID',how='right')
print("\nRight join:")
print(right)
```

```python
outer=pd.merge(students,marks,on='Student_ID',how='outer')
print("\nOuter join:")
print(outer)
```

**Column names differ**

```python
student_info=students.rename(columns={'Student_ID':'ID'})
print("\nStudent information:")
print(student_info)
combined = pd.merge(student_info, marks, left_on='ID', right_on='Student_ID',how='inner')
print("\nCombined data:")
print(combined)
```

**Join using an index**

```python
students_indexed=students.set_index('Student_ID')
marks_indexed=marks.set_index('Student_ID')
joined=students_indexed.join(marks_indexed,how='left')
print("\nStudents (indexed):")
print(students_indexed)
print("\nMarks (indexed):")
print(marks_indexed)
print("\nJoined:")
print(joined)
```

```python
import pandas as pd
data={
    "Dept":["CSE","CSE","ECE","ECE"],
    "Student_ID":[101,102,201,202],
    "Name":["Aman","Riya","Kabir","Neha"],
    "Marks":[85,92,78,88]
}
df=pd.DataFrame(data)
multi_df=df.set_index(["Dept","Student_ID"])
print(multi_df)
multi_df.index.names=["Department","Student_ID"]
print(multi_df)
```

```python
swapped_df=multi_df.swaplevel("Department","Student_ID")
print(swapped_df)
```

**Join by index**

```python
left=pd.DataFrame(
    {"Dept":["A","B","C","D"]},
     index=[101,102,103,104])
right=pd.DataFrame(
    {"Name":["Aman","Riya","Kabir","Neha"]},
    index=[101,102,1033,104])
joined_df=left.join(right)
print(joined_df)
```

**Reshaping and Pivoting**

```python
wide=pd.DataFrame({
    "Student":["Aman","Riya","Kabir","Neha"],
    "Math":[85,92,78,88],
    "Science":[90,89,92,80],
    "English":[88,95,90,85]
})
print(wide)
```

```python
long=pd.melt(
    wide,
    id_vars=["Student"],
    var_name="Subject",
    value_name="Marks"
)
print(long)
```

```python
wide_again=long.pivot(
    index="Student",
    columns="Subject",
    values="Marks"
)
print(wide_again)
```

**Pivot when duplicate combinations exist**

```python
sales=pd.DataFrame({
    "City":["Agra","Agra","Delhi","Delhi","Agra"],
    "Product":["Pen","Pen","Pen","Book","Book"],
    "Sales":[100,150,200,300,250]
   })
print(sales)
```

```python
data=pd.DataFrame({
    "Dept":["CSE","CSE","ECE","ECE","CSE"],
    "Result":["Pass","Fail","Pass","Pass","Pass"]
})
table=pd.crosstab(data["Dept"],data["Result"])
print(table)
```

```python
print(pd.crosstab(data["Dept"],data["Result"],margins=True))
```

```python
print(pd.crosstab(data["Dept"],data["Result"],normalize="index"))
```
